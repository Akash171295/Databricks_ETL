```python
%sql
with first_order as (
    select customer_id,min(order_id) as first_order_id
    from orders
    group by customer_id
)
select sum(price * quantity) as total_revenue
from first_order fo
join order_items oi
on fo.first_order_id = oi.order_id
```

```python
%sql
WITH first_orders AS (
    SELECT 
        customer_id,
        MIN(order_id) AS first_order_id
    FROM orders
    GROUP BY customer_id
)
SELECT 
    SUM(oi.quantity * oi.price) AS new_customer_revenue
FROM first_orders fo
left JOIN order_items oi
ON fo.first_order_id = oi.order_id;
```

```python
%sql
WITH ranked_orders AS (
    SELECT *,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id 
            ORDER BY order_date
        ) AS rn
    FROM orders
)
SELECT 
    SUM(oi.quantity * oi.price) AS new_customer_revenue
FROM ranked_orders r
inner JOIN order_items oi
ON r.order_id = oi.order_id
WHERE r.rn = 1;
```

```python
%sql
with cte as(
select sale_id,product_name,
sum(price*quantity) as product_revenue
from sales
group by sale_id,product_name
),
cum_total as (
  select sale_id,product_name,product_revenue,
  sum(product_revenue) over(order by product_revenue desc) as cumulative_revenue,
  sum(product_revenue) over () as total_revenue
  from cte
)
select sale_id,product_name,product_revenue,
round(cumulative_revenue * 100 /total_revenue,2) as cum_per
from cum_total
where cumulative_revenue * 1.0/ total_revenue < 0.8;
```
